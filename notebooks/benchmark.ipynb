# mini-vllm benchmark

Run the fixed-seed GPT-2 benchmark on one GPU. Set `REPOSITORY_URL` to this repository before running the clone cell. The default sweep uses three trials per configuration and records the GPU model in `results/benchmark.json`.

```python
from pathlib import Path
import os
import subprocess
import sys

REPOSITORY_URL = "https://github.com/mohi-devhub/mini-vllm.git"
REPOSITORY_DIR = Path.cwd() / "mini-vllm"
if not (REPOSITORY_DIR / "pyproject.toml").exists():
    subprocess.run(["git", "clone", REPOSITORY_URL, str(REPOSITORY_DIR)], check=True)
os.chdir(REPOSITORY_DIR)
subprocess.run([sys.executable, "-m", "pip", "install", "-e", ".[benchmark,test]"], check=True)
```

```python
subprocess.run([
    sys.executable, "benchmarks/run.py",
    "--num-requests", "24",
    "--trials", "3",
    "--rates", "2", "4", "8",
    "--output-dir", "results",
], check=True)
```
